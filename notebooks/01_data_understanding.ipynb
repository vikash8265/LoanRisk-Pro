```python
# Features (X) aur Target (y) separate karna

X = df.drop(columns=["LoanID", "Default"])
y = df["Default"]

print("X Shape:", X.shape)
print("y Shape:", y.shape)

print("\nFeatures used for prediction:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("===== TRAIN / TEST SPLIT =====")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)

print("\nDefault = 1 percentage:")
print("Train:", round(y_train.mean() * 100, 2), "%")
print("Test :", round(y_test.mean() * 100, 2), "%")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Numerical columns
numerical_features = [
    "Age",
    "Income",
    "LoanAmount",
    "CreditScore",
    "MonthsEmployed",
    "NumCreditLines",
    "InterestRate",
    "LoanTerm",
    "DTIRatio"
]

# Categorical columns
categorical_features = [
    "Education",
    "EmploymentType",
    "MaritalStatus",
    "HasMortgage",
    "HasDependents",
    "LoanPurpose",
    "HasCoSigner"
]

# Preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

print("Numerical Features  :", len(numerical_features))
print("Categorical Features:", len(categorical_features))
print("Total Original Features:", len(numerical_features) + len(categorical_features))

print("\nPreprocessor created successfully!")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42
        ))
    ]
)

print("Logistic Regression pipeline created successfully!")
```

```python
logistic_model.fit(X_train, y_train)

print("Logistic Regression training completed!")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Improved preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

# Logistic Regression pipeline
logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=42
        ))
    ]
)

# Train again
logistic_model.fit(X_train, y_train)

print("Scaled Logistic Regression training completed successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Predictions
y_pred = logistic_model.predict(X_test)

# Probability of Default = 1
y_prob = logistic_model.predict_proba(X_test)[:, 1]

print("===== LOGISTIC REGRESSION EVALUATION =====")

print("Accuracy :", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred), 4))
print("Recall   :", round(recall_score(y_test, y_pred), 4))
print("F1 Score :", round(f1_score(y_test, y_pred), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_prob), 4))

print("\n===== CONFUSION MATRIX =====")
print(confusion_matrix(y_test, y_pred))

print("\n===== CLASSIFICATION REPORT =====")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline

decision_tree_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", DecisionTreeClassifier(
            max_depth=10,
            min_samples_split=20,
            min_samples_leaf=10,
            class_weight="balanced",
            random_state=42
        ))
    ]
)

decision_tree_model.fit(X_train, y_train)

print("Decision Tree training completed successfully!")
```

```python
# Training data ke andar validation set banana

X_train_final, X_val, y_train_final, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

print("===== TRAIN / VALIDATION / TEST =====")

print("Training   :", X_train_final.shape, y_train_final.shape)
print("Validation :", X_val.shape, y_val.shape)
print("Test       :", X_test.shape, y_test.shape)

print("\nDefault = 1 percentage:")
print("Training   :", round(y_train_final.mean() * 100, 2), "%")
print("Validation :", round(y_val.mean() * 100, 2), "%")
print("Test       :", round(y_test.mean() * 100, 2), "%")
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import roc_auc_score

gb_configs = [
    {"learning_rate": 0.05, "max_iter": 200, "max_leaf_nodes": 15},
    {"learning_rate": 0.05, "max_iter": 300, "max_leaf_nodes": 31},
    {"learning_rate": 0.08, "max_iter": 200, "max_leaf_nodes": 31},
    {"learning_rate": 0.10, "max_iter": 200, "max_leaf_nodes": 31},
    {"learning_rate": 0.08, "max_iter": 300, "max_leaf_nodes": 63},
]

best_auc = -1
best_config = None
best_gb_model = None

print("===== GRADIENT BOOSTING TUNING =====")

for i, config in enumerate(gb_configs, start=1):

    model = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", HistGradientBoostingClassifier(
                learning_rate=config["learning_rate"],
                max_iter=config["max_iter"],
                max_leaf_nodes=config["max_leaf_nodes"],
                random_state=42
            ))
        ]
    )

    model.fit(X_train_final, y_train_final)

    val_prob = model.predict_proba(X_val)[:, 1]
    auc = roc_auc_score(y_val, val_prob)

    print(
        f"{i}/{len(gb_configs)} | "
        f"LR={config['learning_rate']} | "
        f"Iterations={config['max_iter']} | "
        f"Leaves={config['max_leaf_nodes']} | "
        f"Validation AUC={auc:.4f}"
    )

    if auc > best_auc:
        best_auc = auc
        best_config = config
        best_gb_model = model

print("\n===== BEST CONFIGURATION =====")
print("Best Parameters:", best_config)
print("Best Validation ROC-AUC:", round(best_auc, 4))
```

```python
import numpy as np
from sklearn.metrics import precision_score, recall_score, f1_score

# Best model ki validation probabilities
val_prob = best_gb_model.predict_proba(X_val)[:, 1]

best_threshold = None
best_f1 = -1

print("Threshold | Precision | Recall | F1")
print("-----------------------------------")

for threshold in np.arange(0.10, 0.51, 0.025):

    val_pred = (val_prob >= threshold).astype(int)

    precision = precision_score(y_val, val_pred, zero_division=0)
    recall = recall_score(y_val, val_pred)
    f1 = f1_score(y_val, val_pred)

    print(
        f"{threshold:.3f}     | "
        f"{precision:.4f}    | "
        f"{recall:.4f} | "
        f"{f1:.4f}"
    )

    if f1 > best_f1:
        best_f1 = f1
        best_threshold = threshold

print("\n===== BEST VALIDATION THRESHOLD =====")
print("Best Threshold:", round(best_threshold, 3))
print("Best Validation F1:", round(best_f1, 4))
```

```python
import json
import joblib
import pandas as pd

from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.ensemble import HistGradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

from preprocess import create_preprocessor


PROJECT_ROOT = Path(__file__).resolve().parents[1]

DATA_DIR = PROJECT_ROOT / "data"
MODELS_DIR = PROJECT_ROOT / "models"

MODELS_DIR.mkdir(exist_ok=True)


def find_dataset():

    matches = [
        p
        for p in DATA_DIR.rglob("Loan_default.csv")
        if p.is_file()
    ]

    if not matches:
        raise FileNotFoundError(
            "Loan_default.csv not found inside data folder."
        )

    return matches[0]


def main():

    print("=" * 60)
    print("LOANRISK PRO - MODEL TRAINING")
    print("=" * 60)

    dataset_path = find_dataset()

    print("\nDataset:")
    print(dataset_path)

    df = pd.read_csv(dataset_path)

    print("\nDataset shape:", df.shape)

    # Feature / target separation
    X = df.drop(columns=["LoanID", "Default"])
    y = df["Default"]

    # Same deterministic split
    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    print("\nTraining rows:", len(X_train))
    print("Testing rows :", len(X_test))

    preprocessor = create_preprocessor()

    model = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "model",
                HistGradientBoostingClassifier(
                    learning_rate=0.05,
                    max_iter=200,
                    max_leaf_nodes=15,
                    random_state=42
                )
            )
        ]
    )

    print("\nTraining model...")

    model.fit(
        X_train,
        y_train
    )

    print("Training complete.")

    threshold = 0.175

    probabilities = model.predict_proba(X_test)[:, 1]

    predictions = (
        probabilities >= threshold
    ).astype(int)

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions
    )

    f1 = f1_score(
        y_test,
        predictions
    )

    auc = roc_auc_score(
        y_test,
        probabilities
    )

    cm = confusion_matrix(
        y_test,
        predictions
    )

    print("\n===== MODEL RESULTS =====")

    print(
        "Accuracy :",
        round(accuracy, 4)
    )

    print(
        "Precision:",
        round(precision, 4)
    )

    print(
        "Recall   :",
        round(recall, 4)
    )

    print(
        "F1 Score :",
        round(f1, 4)
    )

    print(
        "ROC-AUC  :",
        round(auc, 4)
    )

    print("\nConfusion Matrix:")
    print(cm)

    # Save model
    model_path = (
        MODELS_DIR /
        "loanrisk_model.joblib"
    )

    joblib.dump(
        model,
        model_path
    )

    metadata = {

        "model_name":
            "HistGradientBoostingClassifier",

        "threshold":
            threshold,

        "learning_rate":
            0.05,

        "max_iter":
            200,

        "max_leaf_nodes":
            15,

        "accuracy":
            float(accuracy),

        "precision":
            float(precision),

        "recall":
            float(recall),

        "f1_score":
            float(f1),

        "roc_auc":
            float(auc),

        "dataset_type":
            "Synthetic / simulated loan default dataset"
    }

    metadata_path = (
        MODELS_DIR /
        "model_metadata.json"
    )

    with open(
        metadata_path,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            metadata,
            f,
            indent=4
        )

    print("\nModel saved:")
    print(model_path)

    print("\nMetadata saved:")
    print(metadata_path)


if __name__ == "__main__":
    main()
```
